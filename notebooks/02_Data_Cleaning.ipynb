## Import Required Library

Pandas is used to manipulate and clean structured datasets.

```python
import pandas as pd
import os
```

## Load Dataset


Load the raw dataset for cleaning.

```python
df = pd.read_csv("../data/raw/superstore.csv")

# Display dataset dimensions
print(f"Dataset Shape: {df.shape}")
```

## Check Data Types

```python
df.dtypes
```

## Convert Date Columns

The `Order Date` and `Ship Date` columns are currently stored as text (`object`).

To perform time-based analysis such as yearly, monthly, or quarterly sales trends, these columns must be converted to the `datetime` data type.

The dataset uses the **DD/MM/YYYY** date format, so `dayfirst=True` is used during conversion.

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)
```

## Verify Data Types

```python
df.dtypes
```

## Check Missing Values

```python
df.isnull().sum()
```

## Check Duplicate Records

```python
df.duplicated().sum()
```

## Generate Descriptive Statistics

```python
df.describe()
```

## Feature Engineering

Create additional columns that will be useful for time-based analysis.

```python
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month_name()
df["Order Quarter"] = df["Order Date"].dt.quarter
```

## Preview Cleaned Dataset

```python
df.head()
```

## Export Cleaned Dataset

```python
df.to_csv("../data/cleaned/superstore_cleaned.csv", index=False)
```

```python
output_path = "../data/cleaned/superstore_cleaned.csv"

if os.path.exists(output_path):
    print("✅ Cleaned dataset saved successfully!")
    print(f"Location: {output_path}")
else:
    print("❌ File was not saved.")
```
