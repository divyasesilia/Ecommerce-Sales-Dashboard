## Import Required Libraries

Import the libraries required for data analysis and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Load the Cleaned Dataset

Load the cleaned dataset created in Notebook 02.

```python
df = pd.read_csv("../data/cleaned/superstore_cleaned.csv")

print(df.shape)

df.head()
```

# Sales Analysis

Sales analysis measures the company's overall revenue performance and provides a baseline for evaluating business growth.

## Business Question

What are the total sales generated by the company?

```python
total_sales = df["Sales"].sum()

print(f"Total Sales: ${total_sales:,.2f}")
```

## Business Insight

The total revenue generated by the company is approximately **$2.30 million**.

This metric provides an overall view of business performance and serves as the baseline for further analysis.

## Business Recommendation

- Continue investing in high-performing products.
- Increase customer retention initiatives.
- Monitor revenue growth over time.

## Business Question

What is the average sales value per transaction?

```python
average_sales = df["Sales"].mean()

print(f"Average Sales: ${average_sales:.2f}")
```

## Business Insight

The average sales value per transaction is approximately **$229.86**.

This metric helps identify unusually high or low-value transactions.

## Business Recommendation

Increase the average order value by promoting bundled products and cross-selling complementary items.

## Sales Distribution

Visualize the distribution of sales values to understand how transactions are spread across different sales amounts.

```python
plt.figure(figsize=(8,5))

plt.hist(df["Sales"], bins=30)

plt.title("Distribution of Sales")

plt.xlabel("Sales")

plt.ylabel("Frequency")

plt.show()
```

## Business Insight

Most sales transactions are relatively small, while a small number of orders contribute very high sales values.

This indicates a right-skewed distribution, which is common in retail sales datasets.

## Business Recommendation

Identify the characteristics of high-value orders and develop strategies to increase their frequency through targeted marketing and premium product offerings.

# Profit Analysis

Profit analysis helps evaluate the company's financial performance beyond total sales.

A business may generate high sales but still experience low or negative profits due to discounts, costs, or product mix.

## Business Question

What is the total profit generated by the company?

```python
total_profit = df["Profit"].sum()

print(f"Total Profit: ${total_profit:,.2f}")
```

## Business Insight

The company generated approximately **$286,397** in total profit.

Comparing total sales with total profit provides an overall understanding of business profitability.

## Business Recommendation

Review operational costs and pricing strategies to further improve profitability while maintaining sales performance.

## Business Question

What is the average profit earned per order?

```python
average_profit = df["Profit"].mean()

print(f"Average Profit: ${average_profit:.2f}")
```

## Business Insight

On average, each transaction generates approximately **$28.66** in profit.

This metric helps evaluate the overall profitability of individual orders.

## Business Recommendation

Focus on increasing the average profit per order by reducing unnecessary discounts and promoting higher-margin products.

## Profit Distribution

Visualize the distribution of profit values to identify profitable and loss-making transactions.

```python
plt.figure(figsize=(8,5))

plt.hist(df["Profit"], bins=30)

plt.title("Distribution of Profit")

plt.xlabel("Profit")

plt.ylabel("Frequency")

plt.show()
```

## Business Insight

The profit distribution includes both positive and negative values.

Negative profits indicate loss-making orders, while the right side of the distribution represents highly profitable transactions.

This suggests that not every sale contributes positively to the company's profitability.

## Business Recommendation

Investigate loss-making orders to identify the root causes and implement corrective actions to improve overall profitability.

# Category Analysis

Category analysis helps identify which product categories contribute the most to sales and profit.

Understanding category performance enables businesses to focus on high-performing product lines and improve low-performing ones.

## Business Question

Which product categories generate the highest sales?

```python
category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(category_sales)
```

## Business Insight

Technology generates the highest sales, followed by Furniture and Office Supplies.

This indicates that Technology products are the company's primary revenue driver.

## Business Recommendation

Continue investing in Technology products through inventory expansion and targeted marketing campaigns while improving sales strategies for lower-performing categories.

## Sales by Category

Visualize total sales generated by each product category.

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")

plt.xlabel("Category")

plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

## Business Insight

Technology consistently outperforms the other categories in terms of sales.

Management should continue investing in high-performing product categories while identifying opportunities to improve lower-performing categories.

## Business Recommendation

Maintain investment in Technology while evaluating pricing and operational costs within Furniture.

## Business Question

Which product category generates the highest profit?

```python
category_profit = (
    df.groupby("Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

print(category_profit)
```

## Business Insight

Comparing category profit with category sales helps determine whether high sales also translate into high profitability.

A category with strong sales but weak profit may require pricing or cost optimization.

## Business Recommendation

Increase investment in high-margin Technology products while reviewing cost structures in less profitable categories.

## Profit by Category

Compare total profit across product categories.

```python
plt.figure(figsize=(8,5))

category_profit.plot(kind="bar")

plt.title("Total Profit by Category")

plt.xlabel("Category")

plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.show()
```

## Business Insight

This visualization highlights the most profitable product categories.

Business leaders can use this information to prioritize inventory, marketing, and investment decisions.

## Business Recommendation

Focus on improving profit margins in Furniture through optimized pricing, supplier negotiations, and discount management.

# Sub-Category Analysis

Sub-category analysis provides a more detailed view of product performance.

While categories provide a high-level overview, sub-categories help identify the specific product groups contributing to sales and profit.

## Business Question

Which sub-categories generate the highest sales?

```python
subcategory_sales = (
    df.groupby("Sub-Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(subcategory_sales)
```

## Business Insight

The highest-selling sub-categories contribute a significant portion of the company's total revenue.

These products should receive continued attention in inventory planning and marketing strategies.

## Business Recommendation

Ensure sufficient inventory levels and prioritize marketing campaigns for these high-performing products.

## Sales by Sub-Category

```python
plt.figure(figsize=(12,6))

subcategory_sales.plot(kind="bar")

plt.title("Total Sales by Sub-Category")

plt.xlabel("Sub-Category")

plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

## Business Insight

Sales are concentrated in only a few sub-categories, indicating that several products contribute disproportionately to total revenue.

## Business Recommendation

Continue investing in top-performing products while exploring strategies to improve the sales of underperforming sub-categories.

## Business Question

Which sub-categories generate the highest profit?

```python
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

print(subcategory_profit)
```

## Business Insight

Copiers generate the highest profit despite not having the highest sales.

Tables and Bookcases generate relatively low or negative profit despite contributing to revenue.

## Business Recommendation

Review pricing strategies and discount policies for Tables and Bookcases to improve profitability while maintaining sales.

## Profit by Sub-Category

The chart below compares profit across all sub-categories.

```python
plt.figure(figsize=(12,6))

subcategory_profit.plot(kind="bar")

plt.title("Total Profit by Sub-Category")

plt.xlabel("Sub-Category")

plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.show()
```

## Business Insight

Several sub-categories generate losses even though they contribute substantial sales.

This highlights opportunities for improving pricing, controlling costs, or reducing excessive discounts.

## Business Recommendation

Conduct a detailed profitability review of Tables and Bookcases to identify the primary causes of losses and implement corrective actions.

# Discount Analysis

Discount analysis evaluates how discounts influence company profitability.

## Business Question

How do discounts affect profit?

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Discount"], df["Profit"])

plt.title("Discount vs Profit")

plt.xlabel("Discount")

plt.ylabel("Profit")

plt.show()
```

## Business Insight

Higher discounts are generally associated with lower profits and, in many cases, financial losses.

This suggests that excessive discounting negatively impacts profitability.

# Monthly Sales Trend

Monthly sales analysis identifies seasonal patterns and changes in customer demand throughout the year.

## Business Question

How do sales change across different months?

```python
monthly_sales = (
    df.groupby("Order Month")["Sales"]
      .sum()
)

print(monthly_sales)
```

## Business Insight

Monthly sales analysis helps identify peak and slow business periods.

Understanding seasonality supports inventory planning and promotional strategies.

## Monthly Sales Trend

The chart below illustrates total sales for each month.

```python
plt.figure(figsize=(10,5))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.grid(True)

plt.show()
```

## Business Insight

Sales fluctuate throughout the year, with noticeable peak months.

The company should prepare inventory and marketing campaigns ahead of high-demand periods.
