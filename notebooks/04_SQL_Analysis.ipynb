## Import Required Libraries

Import the libraries required to connect Python with SQLite and execute SQL queries.

```python
import pandas as pd
import sqlite3
```

## Load the Cleaned Dataset

Load the cleaned dataset created in Notebook 02.

```python
df = pd.read_csv("../data/cleaned/superstore_cleaned.csv")

print(f"Dataset Shape: {df.shape}")

df.head()
```

## Create SQLite Database

Create an in-memory SQLite database and load the cleaned dataset into a SQL table.

```python
conn = sqlite3.connect(":memory:")

df.to_sql(
    "superstore",
    conn,
    index=False,
    if_exists="replace"
)

print("SQLite database created successfully.")
print("Table 'superstore' is ready for SQL analysis.")
```

## SQL Table Preview

Display the first five records from the SQLite table to verify that the data has been loaded correctly.

```python
query = """
SELECT *
FROM superstore
LIMIT 5;
"""

pd.read_sql(query, conn)
```

# Data Validation

Before performing SQL analysis, verify the dataset structure and quality.

```python
query = """
SELECT
    COUNT(*) AS Total_Rows,
    ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore;
"""

pd.read_sql(query, conn)
```

## Validation Result

- The cleaned dataset has been successfully loaded into the SQLite database.
- The total number of records matches the cleaned dataset.
- The SQL table is ready for business analysis.


# Sales Analysis

Sales analysis measures the company's overall revenue performance using SQL queries.

## Business Question

What are the total sales generated by the company?

```python
query = """
SELECT
ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore;
"""

pd.read_sql(query, conn)
```

## Business Insight

The company generated approximately **$2.30 million** in total sales.

This represents the total revenue generated from all customer transactions.

## Business Recommendation

Management should monitor total sales regularly to evaluate business growth and measure the effectiveness of sales strategies.

## Business Question

What is the average sales value per transaction?

```python
query = """
SELECT
ROUND(AVG(Sales),2) AS Average_Sales
FROM superstore;
"""

pd.read_sql(query, conn)
```

## Business Insight

The average sales value per transaction is approximately **$229.86**.

This metric helps identify changes in customer purchasing behavior.

## Business Recommendation

Increasing average order value through product bundling and upselling can improve revenue.

# Profit Analysis

Profit analysis evaluates the company's financial performance using SQL queries.

## Business Question

What is the total profit generated by the company?

This indicates that the business is profitable overall.

## Business Recommendation

Focus on improving profit margins while maintaining sales growth.

## Business Question

## Business Recommendation

# Category Analysis

Category analysis compares the sales and profitability of major product categories.

This helps identify which categories contribute the most to the company's overall business performance.

## Business Question

Which product category generates the highest sales?

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

## Business Insight

Technology generates the highest total sales, followed by Furniture and Office Supplies.

This indicates that Technology products are the primary driver of the company's revenue.

## Business Recommendation

Continue investing in Technology products while identifying opportunities to improve sales in the Furniture and Office Supplies categories.

## Business Question

Which product category generates the highest profit?

```python
query = """
SELECT
    Category,
    ROUND(SUM(Profit),2) AS Total_Profit
FROM superstore
GROUP BY Category
ORDER BY Total_Profit DESC;
"""

pd.read_sql(query, conn)
```

## Business Insight

Technology is the most profitable category, while Furniture generates comparatively lower profit despite strong sales.

This suggests that high sales do not always translate into high profitability.

## Business Recommendation

Improve profitability in the Furniture category by reviewing pricing strategies, procurement costs, and discount policies.

# Sub-Category Analysis

Sub-category analysis provides a detailed view of product performance within each category.

It helps identify the highest-performing and lowest-performing product groups.

## Business Question

Which sub-categories generate the highest sales?

```python
query = """
SELECT
    [Sub-Category],
    ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY [Sub-Category]
ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, conn)
```

## Business Insight

Phones, Chairs, and Storage are among the highest-selling sub-categories.

These products contribute a significant share of the company's overall revenue.

## Business Recommendation

Maintain sufficient inventory levels and prioritize marketing efforts for high-selling sub-categories to sustain revenue growth.

## Business Question

Which sub-categories generate the highest profit?

```python
query = """
SELECT
    [Sub-Category],
    ROUND(SUM(Profit),2) AS Total_Profit
FROM superstore
GROUP BY [Sub-Category]
ORDER BY Total_Profit DESC;
"""

pd.read_sql(query, conn)
```

## Business Insight

Copiers generate the highest profit, while Tables and Bookcases produce comparatively lower profits.

Although Tables and Bookcases contribute to sales, their profitability is significantly lower than other sub-categories, indicating potential issues with pricing, discounts, or operational costs.

## Business Recommendation

Review pricing strategies, supplier costs, and discount policies for Tables and Bookcases to improve profitability while maintaining sales performance.

# Customer Analysis

Customer analysis identifies the company's highest-value customers based on sales.

Understanding customer contribution supports customer retention and loyalty initiatives.

## Business Question

Who are the top 10 customers by total sales?

```python
query = """
SELECT
    [Customer Name],
    ROUND(SUM(Sales),2) AS Total_Sales,
    ROUND(SUM(Profit),2) AS Total_Profit
FROM superstore
GROUP BY [Customer Name]
ORDER BY Total_Sales DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```

## Business Insight

A small group of customers contributes a significant portion of total sales.

These customers represent valuable long-term business relationships.

## Business Recommendation

Strengthen customer loyalty through personalized services, targeted promotions, and retention programs for high-value customers.

# Discount Analysis

Discount analysis examines how pricing discounts influence business profitability.

Understanding this relationship helps optimize promotional strategies.

## Business Question

How does discount affect average profit?

```python
query = """
SELECT
    Discount,
    ROUND(AVG(Profit),2) AS Average_Profit
FROM superstore
GROUP BY Discount
ORDER BY Discount;
"""

pd.read_sql(query, conn)
```

## Business Insight

Higher discount levels generally result in lower average profit.

Several heavily discounted transactions generate losses despite contributing to sales.

## Business Recommendation

Review discount policies carefully to balance increased sales volume with sustainable profitability.

# Monthly Sales Trend

Monthly sales trend analysis identifies seasonal purchasing behavior throughout the year.

This helps businesses plan inventory and promotional campaigns.

## Business Question

What are the monthly sales trends?

```python
query = """
SELECT
    [Order Year],
    [Order Month],
    ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY
    [Order Year],
    [Order Month]
ORDER BY
    [Order Year],
    MIN([Order Date]);
"""

pd.read_sql(query, conn)
```

## Business Insight

Sales fluctuate across different months, indicating seasonal demand patterns.

Certain months consistently generate higher revenue than others.

## Business Recommendation

Plan marketing campaigns, inventory replenishment, and staffing based on expected seasonal demand.

# Yearly Sales Trend

Yearly sales trend analysis evaluates long-term business growth.

Monitoring annual performance supports strategic planning and forecasting.

## Business Question

How have sales changed over the years?

```python
query = """
SELECT
    [Order Year],
    ROUND(SUM(Sales),2) AS Total_Sales
FROM superstore
GROUP BY
    [Order Year]
ORDER BY
    [Order Year];
"""

pd.read_sql(query, conn)
```

## Business Insight

The company demonstrates an overall upward trend in annual sales, indicating positive business growth over time.

Monitoring yearly performance helps assess long-term business success.

## Business Recommendation

Continue monitoring annual sales trends to support strategic planning, budgeting, and future business expansion.
