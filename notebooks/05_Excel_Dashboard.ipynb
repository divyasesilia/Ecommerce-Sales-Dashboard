# KPI Cards

The dashboard displays the following Key Performance Indicators (KPIs):

- Total Sales
- Total Profit
- Average Sales
- Average Profit

## Business Insight

The KPI cards provide an instant overview of overall business performance.

Users can quickly evaluate revenue, profitability, and average transaction value without exploring detailed reports.

## Business Recommendation

Management should monitor these KPIs regularly to track business performance and support data-driven decision-making.

# Sales by Category

A Pivot Chart compares sales across product categories.

## Business Insight

Technology generates the highest sales, followed by Furniture and Office Supplies.

## Business Recommendation

Continue investing in Technology products while identifying growth opportunities for the remaining categories.

# Profit by Category

A Pivot Chart compares profitability across product categories.

## Business Insight

Technology generates the highest profit, while Furniture has comparatively lower profitability.

## Business Recommendation

Review pricing strategies and discount policies for Furniture products.

# Sales by Region

A Pivot Chart compares regional sales performance.

## Business Insight

The West region contributes the highest revenue.

## Business Recommendation

Replicate successful sales strategies from the West region in lower-performing regions.

# Profit by Sub-Category

A Pivot Chart compares profitability across product groups.

## Business Insight

Copiers are the most profitable products.

Tables and Bookcases generate comparatively lower profits.

## Business Recommendation

Review supplier costs and discount policies for Tables and Bookcases.

# Monthly Sales Trend

A line chart illustrates monthly sales performance.

## Business Insight

Sales fluctuate throughout the year, indicating seasonal purchasing behavior.

## Business Recommendation

Plan inventory, staffing, and marketing campaigns based on seasonal demand.

# Top Customers

A Pivot Table displays the company's highest-value customers based on total sales.

## Business Insight

A relatively small group of customers contributes a significant portion of total revenue.

## Business Recommendation

Develop customer loyalty programs and personalized offers to retain high-value customers.

# Interactive Dashboard Features

The dashboard includes slicers that allow users to filter the report dynamically.

Available filters include:

- Region
- Category
- Segment
- Ship Mode
- Year

# Executive Summary

## Key Findings

- Technology generates the highest sales and profit.
- Copiers are the most profitable sub-category.
- Tables and Bookcases have comparatively lower profitability.
- The West region performs best.
- Sales show seasonal trends.
- A small number of customers contribute significantly to total revenue.

## Business Recommendations

- Increase investment in high-performing categories.
- Optimize pricing for low-profit products.
- Improve customer retention strategies.
