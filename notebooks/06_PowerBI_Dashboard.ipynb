- Total Profit
- Average Sales
- Average Profit

## Business Insight

The KPI cards provide an instant overview of business performance.

Users can quickly evaluate revenue, profitability, and average transaction value from a single dashboard.

## Business Recommendation

Management should regularly monitor these KPIs to track overall business performance and identify significant changes over time.

# Sales by Category

A clustered bar chart compares total sales across product categories.

## Business Insight

Technology generates the highest sales, followed by Furniture and Office Supplies.

## Business Recommendation

Continue investing in Technology products while identifying growth opportunities in the remaining categories.

# Profit by Category

A clustered column chart compares profitability across product categories.

## Business Insight

Technology generates the highest profit, while Furniture has comparatively lower profitability.

## Business Recommendation

Review pricing strategies and discount policies for Furniture products to improve profitability.

# Sales by Region

A horizontal bar chart compares total sales performance across regions.

## Business Insight

The West region contributes the highest revenue among all regions.

## Business Recommendation

Replicate successful sales strategies from the West region in lower-performing regions.

# Profit by Sub-Category

A bar chart compares profitability across product sub-categories.

## Business Insight

Copiers generate the highest profit, while Tables and Bookcases contribute comparatively lower profits.

## Business Recommendation

Review supplier costs, pricing strategies, and discount policies for low-profit sub-categories.

# Monthly Sales Trend

A line chart illustrates monthly sales performance throughout the year.

## Business Insight

Sales fluctuate throughout the year, indicating seasonal purchasing behavior.

## Business Recommendation

Plan inventory, staffing, and marketing campaigns based on seasonal demand patterns.

# Top 10 Customers

A bar chart displays the company's highest-value customers based on total sales.

## Business Insight

A relatively small group of customers contributes a significant portion of total revenue.

## Business Recommendation

Develop customer loyalty programs and personalized offers to retain high-value customers.
